```python
import pandas as pd 
rdv=pd.read_csv("student_performance_dataset.csv")
rdv.loc[0,'gender'] # loc is the label based selection operator which gives the data on the row whith the column name
```

```python
# Even multiple rows also can be selected using the label based selection operator loc 
rdv.loc[:,['gender','study_time_hours','attendance_percent']]
```

```python
# Index Manipulation
rdv.set_index("student_id")# This is used to keep the index that is immuatble same like the primary key so that the data can be related with that entity
```

```python
# Conditional selection: This operation prints only the rows if the coniditoin is met,
rdv.loc[rdv.attendance_percent>85] 
```
