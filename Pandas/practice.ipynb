```python
import numpy as np 
X,Y =np.loadtxt("pizza.txt",skiprows=1,unpack=True)
def predict(X,w):
    return X*w
print(f"The number of pizza's sold in the branch:{predict(10,2.1)}")
def loss(X,Y,w):
    return np.average((predict(X,w)-Y)**2)
def train(X,Y,it,lr):
    w=0
    for i in range(it):
        cl=loss(X,Y,w)
        if loss(X,Y,w+lr)<cl:
            w+=lr
        elif loss(X,Y,w-lr)<cl:
            w-=lr
        else:
            return w
w=train(X,Y,it=10000,lr=0.01)
print(f"Predication: {20} => {predict(20,w)}")
import matplotlib.pyplot as plt
plt.axis([0,50,0,50])
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel("Reservations",fontsize=14)
plt.ylabel("Pizzas ordered",fontsize=14)
plt.plot(X,Y,"bo")
plt.plot(X,predict(X,w),"g--")
plt.show()

```
