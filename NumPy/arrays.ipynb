# Topic: Creation of the array
*The list is taken as the parameter in the np.array() function and then it is converted into array*
### Arrays are of multiple dimensions in Python 1D,2D,3D,.....ND
* np.zeros() to create the array with 0's 
* np.arange() to create the array from the 0 to the parameterized index
* np.arange(1,6) when two parameter is given it is taken as starting and ending
* np.arange(1,6,2) when three parameters are given then it is taken as the starting and the ending and the step so that the array skips the numbers by the steps

```python
import numpy as np
import random
a=np.array([1,2,3])
print(a)
print(repr(a))
# Two dimensional array
twod=np.array([[1,2,3],[4,5,6]])
print("The two dimensional array is\n",twod)
zero=np.zeros((4,3)) # Take the shape in the parameter to give the array 
print(zero)
ran=np.arange(6)
ran1=np.arange(0,6) # The start and the end parameter is given
ran2=np.arange(0,6,2) # This gives the array with the step of 2
print(ran1)
print(ran2)
print(ran)
rand=np.random.random((2,4)) # Takes the shape in the parameter to give the array
print(rand)
```

# Type Casting:
*This is the method in which we tell numpy to convert the type of the data to an other type*
### This is done using astype method

```python
import numpy as np 
boolean_array=np.array([[True,False],[False,False]],dtype=np.bool)
typeCastedArray=boolean_array.astype(np.int32)
print(repr(typeCastedArray))
(typeCastedArray).dtype
```

# Type Coercion
*If the python list that has multiple data types is changed into an array it would be converted to an array of the same data type*


* *If the list has a string the array will be of the datatype of string*
* *If the list has float values with integers then the array will be of the datatype of float*
* *If the list has one integer and boolean value then the array will be of the data type of integers*

#### The reason why it converts into a single data type is since numpy has a lower memory consumption.

```python
import numpy as np
array=np.array(["True","Anirudh",36,8.36])
print(repr(array))
```

### For 2D array's we have to give the row and the column and to access the array element
#### *Note:*
* *If we are giving only a single index it will take it as the row index and give the entire row*
* *To access the entire column elements we should use a colon and then the column index to access the entire column*

```python
import numpy as np 
a=np.array([[9,3,8,0,7,0,7,5,4,4],[9,6,3,2,9,0,3,7,7,4]])
print(a)
print(a[0])
print(a[:,0])
# It printed the entrire first row when the a[0] was used.
```

# Sorting Arrays:
* *The rows of the array are sorted where the axis is mentioned*
* *By default the np.sort() sorts the array in the column axis*

# Types of Axes:
* *There are two types of axes 1. Row axis 2. Column axis*
* *The row axis is denoted by 0*
* *The column axis is denoted by 1*

```python
import numpy as np
a=np.array([[9,3,8,0,7,0,7,5,4,4],[9,6,3,2,9,0,3,7,7,4]])
print("The unosorted array",a)
print("The sorted array",np.sort(a))
```

### TO sort the array in the column axis, we have to give a new attribute known as axis=0
* *The array will be sorted from top to down in the row so that bigger numbers are at the bottom*
* *In column, the array will be sorted from left to right so that the bigger values are on the right side*

```python
import numpy as np
arr=np.array([[9,3,8,0,7,0,7,5,4,4],[9,6,3,2,9,0,3,7,7,4]])
print("The unsorted array")
print(arr)
print("The sorted array")
print(np.sort(arr,axis=0))
print(np.sort(arr,axis=1))
```
