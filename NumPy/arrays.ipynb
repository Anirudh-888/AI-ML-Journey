```python
import numpy as np
arr = np.array([[0,1,2],[3,4,5]], dtype=np.float32)
print(repr(arr))
# The np.array is used to convert the list into the array by allocation of memory continuously.
```

```python
a=np.array([0,1])
b=np.array([9,8])
c=a
print('Array a:{}'.format(repr(a)))
print(f'Array a:{format(repr(a))}')
c[0]=5
print('Array a:{}'.format(repr(a)))
d=b.copy()
d[0]=6
print('Array b:{}'.format(repr(b)))
# If the change is made in a referenced array,it even changes in the parent list.
# So, using copy(), the copy of the parent array is made and when we change the copied array then the parent array remains unchanged.
print('Array d:{}'.format(repr(d)))
# In the copied d array the value is changed and not in the parent b array.
```

```python
# Casting
import numpy as np
arr=np.array([0,1,2]) 
print(arr.dtype)
# This return the type as int64
arr=arr.astype(np.float32)
print(arr.dtype)
# Using astype function the array is typecasted as float32
```
