```python
a=np.array([0,1])
b=np.array([9,8])
c=a
print('Array a:{}'.format(repr(a)))
print(f'Array a:{format(repr(a))}')
c[0]=5
print('Array a:{}'.format(repr(a)))
d=b.copy()
d[0]=6
print('Array b:{}'.format(repr(b)))
# If the change is made in a referenced array,it even changes in the parent list.
# So, using copy(), the copy of the parent array is made and when we change the copied array then the parent array remains unchanged.
print('Array d:{}'.format(repr(d)))
# In the copied d array the value is changed and not in the parent b array.
```

```python
# Casting
import numpy as np
arr=np.array([0,1,2]) 
print(arr.dtype)
# This return the type as int64
arr=arr.astype(np.float32)
print(arr.dtype)
# Using astype function the array is typecasted as float32
```

```python
# Ranged Data 
# Using arange we can create a unidimensional array within that range of integers
import numpy as np 
arr=np.arange(5)
print(repr(arr))
print(arr.dtype)
arr1=np.arange(-1,4)
print(arr1)
arr2=np.arange(-1.5,4,2)
print(repr(arr2))
arr3=np.arange(0,1,0.1)
print(repr(arr3.astype(np.int32)))
# The final output has typecasted the float values to int values
```

```python
# Reshaping the data, the data can be converted into any other forms but all the data from the previous form must be inlcuded
import numpy as np
a=np.arange(8)
print(repr(a))
ra = np.reshape(a, (2, 4))
print(repr(ra))
ra2=np.reshape(a, (-1,2,2))
print(repr(ra2))
```

```python
# To convert the array into a 1D array we can just directly use the flatten method
import numpy as np
a=np.arange(8)
arr=np.reshape(a, (2,4))
flat=arr.flatten()
print(repr(flat))
print(repr(arr))
```

```python
# Transposing: transosing the matrix
# Transposing operation is that changes the row to column and the column to row
import numpy as np 
a=np.arange(8)
arr=np.reshape(a, (4,2))
transposed = np.transpose(arr)
print(repr(arr))
print(repr(transposed))

```
