```python
# To convert the array into a 1D array we can just directly use the flatten method
import numpy as np
a=np.arange(8)
arr=np.reshape(a, (2,4))
flat=arr.flatten()
print(repr(flat))
print(repr(arr))
```

```python
# Transposing: transosing the matrix
# Transposing operation is that changes the row to column and the column to row
import numpy as np 
a=np.arange(8)
arr=np.reshape(a, (4,2))
transposed = np.transpose(arr)
print(repr(arr))
print(repr(transposed))

```

```python
#Arithematic
import numpy as np
arr=np.array([[1,2],[3,4]])
print(repr(arr))
# Addition operation will add the number to all the array elements
print(repr(arr+1))
# Subtraction operation will subtract the number from all the array elements
print(repr(arr-1))
# Multiplication operation multiplies all the array elements
print(repr(arr*2))
# Division operation divides the number from all the array elements
print(repr(arr/2))
# Squaring all the elements of the array
print(repr(arr**2))
# Root all the elements of the array
print(repr(arr**0.5))
```

```python
# At once we can convert many farenheit values to celcuis or viceversa
def f2c(temps):
    return(5/9)*(temps-32)
fahrenheits = np.array([32,-4,14,40])
fahrenheits=f2c(fahrenheits)
print(repr(fahrenheits))
# Always this creates a new array it doesn't change the parent array
```

```python
import numpy as np
arr=np.array([[1,2],[3,4]])
# Raised to the power of e means e power the elements
print(repr(np.exp(arr)))
# Raised to the power of 2 means 2 power the elements
print(repr(np.exp2(arr)))
arr2=np.array([[1,10],[np.e, np.pi]])
# Natural Log
print(repr(np.log(arr2)))
# Log 10 base
print(repr(np.log10(arr2)))
```

```python
import numpy as np
arr=np.array([[1,2],[3,4]])
print(repr(np.power(3,arr)))
# The 3 acts as a base and the elements of the array is taken as the exponent
arr2=np.array([[10.2,4],[3,5]])
print(repr(np.power(arr2,arr)))
# In this operation the arr2 acts as a base and the arr acts as the power
```

```python
# Matrix multiplication
import numpy as np
arr1=np.array([1,2,3])
arr2=np.array([-3,0,10])
print(np.matmul(arr1,arr2))

arr3=np.array([[1,2],[3,4],[5,6]])
arr4=np.array([[-1,0,1],[3,2,-4]])
print(repr(np.matmul(arr3,arr4)))
print(repr(np.matmul(arr4,arr3)))
```
